## Conversion BDB → JSON

Conversion du dictionnaire LSJ (Lidell Scott Jones) selon notre format de données


### Source

- **Fichier** : `lsj.json` tiré de [Perseis-Project](https://www.perseids.org/libraries-tools), lien GitHub: https://github.com/perseids-project/lsj-js/blob/master/vendor/lsj.json

### Structure JSON produite

| Champ | Type | Description |
|---|---|---|
| `dic` | str | Identifiant du dictionnaire |
| `sort_key` | int | Numéro d'ordre séquentiel |
| `strong` | str/null | Numéro Strong (`H24`), `null` si absent |
| `lang` | str | Langue : `"Hebrew"` ou `"Aramaic"` |
| `m` | list | Formes avec voyelles et accents |
| `b` | list | Formes nues sans diacritiques (pour la recherche) |
| `l` | list | Translitérations latines |
| `d` | str | Définition HTML (balise `<p>`) |


```python
import json

with open('lsj-perseis.json', encoding='utf-8') as f:
    lsj_json = json.load(f)

sort_key = 1

lsj = []
for k, v in lsj_json.items():
    lsj.append({
        "dic": "LSJ",
        "sort_key": sort_key,
        "lang": "Greek",
        "m": v["m"],
        "b": v.pop("g"),
        "l": v["l"],
        "d": v["d"],
    })
    sort_key += 1

with open('lsj.json', 'w', encoding='utf-8') as f:
    json.dump(lsj, f, ensure_ascii=False, indent=2)

print(f"file lsj.json converted — {len(lsj)} entries")
```
